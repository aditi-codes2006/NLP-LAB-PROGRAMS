```python
# Program 3A: Stop-word Removal without NLTK

import re

# Sample document
text = """
Natural Language Processing is a branch of Artificial Intelligence.
It helps computers understand and process human language.
NLP is used in many real-world applications.
"""

print("Original Text:")
print(text)

# Define common English stop words
stop_words = {
    "is", "a", "an", "the", "and", "in", "to",
    "of", "it", "for", "on", "are", "this",
    "that", "with", "as", "by", "from"
}

# Convert text to lowercase and extract words
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Remove stop words
filtered_words = [word for word in words if word not in stop_words]

print("\nWords before Stop-word Removal:")
print(words)

print("\nWords after Stop-word Removal:")
print(filtered_words)
```

```python
# Program 3B: Stop-word Removal using NLTK

import nltk

# Download required resources
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Sample document
text = """
Natural Language Processing is a branch of Artificial Intelligence.
It helps computers understand and process human language.
NLP is used in many real-world applications.
"""

print("Original Text:")
print(text)

# Tokenize the text
words = word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words and punctuation
filtered_words = [
    word for word in words
    if word.lower() not in stop_words and word.isalpha()
]

print("\nWords before Stop-word Removal:")
print(words)

print("\nWords after Stop-word Removal:")
print(filtered_words)
```
