```python
# Import Word2Vec from Gensim
from gensim.models import Word2Vec

# Custom corpus
# Each sentence is represented as a list of words
corpus = [
    ["machine", "learning", "is", "interesting"],
    ["machine", "learning", "is", "powerful"],
    ["deep", "learning", "is", "a", "part", "of", "machine", "learning"],
    ["artificial", "intelligence", "is", "the", "future"],
    ["machine", "learning", "and", "artificial", "intelligence", "are", "related"]
]

# Create and train the Word2Vec model
model = Word2Vec(
    sentences=corpus,   # Training data
    vector_size=100,    # Size of word vectors
    window=5,            # Maximum distance between current and predicted word
    min_count=1,        # Include words appearing at least once
    workers=4,          # Number of CPU threads
    sg=1                # 1 = Skip-Gram, 0 = CBOW
)

# Display the vector representation of a word
word = "machine"

print("Vector representation of:", word)
print(model.wv[word])

# Find words most similar to "machine"
print("\nWords similar to 'machine':")

similar_words = model.wv.most_similar("machine", topn=5)

for word, similarity in similar_words:
    print(word, ":", similarity)
```
