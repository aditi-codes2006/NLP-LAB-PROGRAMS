```python
import math

doc1 = "machine learning is interesting"
doc2 = "machine learning is useful"

words = list(set(doc1.split() + doc2.split()))

vector1 = [doc1.split().count(word) for word in words]
vector2 = [doc2.split().count(word) for word in words]

dot_product = sum(a * b for a, b in zip(vector1, vector2))

magnitude1 = math.sqrt(sum(a * a for a in vector1))
magnitude2 = math.sqrt(sum(b * b for b in vector2))

cosine = dot_product / (magnitude1 * magnitude2)

print("Cosine Similarity:", cosine)
```
