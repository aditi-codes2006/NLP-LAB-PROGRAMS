```python
# Install:
# pip install gensim numpy

import gensim.downloader as api
import numpy as np

# Load pre-trained GloVe model
glove = api.load("glove-wiki-gigaword-50")

# Display vector of a word
word = "computer"

print("Vector representation of", word)
print(glove[word])

# Find similar words
print("\nSimilar words:")
similar_words = glove.most_similar(word, topn=5)

for w, score in similar_words:
    print(w, ":", score)

# Sentence vector
sentence = "machine learning is useful"

words = sentence.lower().split()

valid_words = [word for word in words if word in glove]

sentence_vector = np.mean(
    [glove[word] for word in valid_words],
    axis=0
)

print("\nSentence Vector:")
print(sentence_vector)
```
