```python
# ============================================================
# PROGRAM 20
# INFORMATION EXTRACTION
# ============================================================

import spacy
import re

# ------------------------------------------------------------
# Load spaCy English model
# ------------------------------------------------------------

nlp = spacy.load("en_core_web_sm")

# ------------------------------------------------------------
# Input document
# ------------------------------------------------------------

text = """
Aditi Mishra joined Google in Delhi on 5 October 2026.
She can be contacted at aditi@example.com.
Her phone number is 9876543210.
Google is headquartered in Mountain View.
"""

# ------------------------------------------------------------
# Process document
# ------------------------------------------------------------

doc = nlp(text)

# ------------------------------------------------------------
# Named Entity Recognition
# ------------------------------------------------------------

print("================================")
print("NAMED ENTITIES")
print("================================")

for entity in doc.ents:

    print(
        "Entity:",
        entity.text,
        "| Type:",
        entity.label_
    )

# ------------------------------------------------------------
# Extract Email
# ------------------------------------------------------------

emails = re.findall(
    r'[\w\.-]+@[\w\.-]+\.\w+',
    text
)

print("\n================================")
print("EMAIL ADDRESSES")
print("================================")

for email in emails:
    print(email)

# ------------------------------------------------------------
# Extract Phone Number
# ------------------------------------------------------------

phones = re.findall(
    r'\b\d{10}\b',
    text
)

print("\n================================")
print("PHONE NUMBERS")
print("================================")

for phone in phones:
    print(phone)
```
