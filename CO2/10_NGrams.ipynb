```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is interesting"
]

# -----------------------------
# 1. Unigram (1-gram)
# -----------------------------
vectorizer = CountVectorizer(ngram_range=(1, 1))
unigram = vectorizer.fit_transform(documents)

print("Unigrams:")
print(vectorizer.get_feature_names_out())
print(unigram.toarray())


# -----------------------------
# 2. Bigram (2-gram)
# -----------------------------
vectorizer = CountVectorizer(ngram_range=(2, 2))
bigram = vectorizer.fit_transform(documents)

print("\nBigrams:")
print(vectorizer.get_feature_names_out())
print(bigram.toarray())


# -----------------------------
# 3. Trigram (3-gram)
# -----------------------------
vectorizer = CountVectorizer(ngram_range=(3, 3))
trigram = vectorizer.fit_transform(documents)

print("\nTrigrams:")
print(vectorizer.get_feature_names_out())
print(trigram.toarray())


# -----------------------------
# 4. Unigram + Bigram + Trigram
# -----------------------------
vectorizer = CountVectorizer(ngram_range=(1, 3))
ngrams = vectorizer.fit_transform(documents)

print("\nUnigram + Bigram + Trigram:")
print(vectorizer.get_feature_names_out())
print(ngrams.toarray())
```
