```python
# Install required packages (pyemd is needed for WMD):
!pip install gensim nltk pyemd

import nltk
from gensim.models import Word2Vec

nltk.download('punkt')

# Training corpus
corpus = [
    ["the", "cat", "is", "on", "the", "mat"],
    ["the", "dog", "is", "on", "the", "mat"],
    ["the", "cat", "likes", "milk"],
    ["the", "dog", "likes", "meat"],
    ["a", "kitten", "is", "a", "small", "cat"]
]

# Train Word2Vec model
model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

# Two documents
doc1 = ["cat", "likes", "milk"]
doc2 = ["kitten", "likes", "milk"]
doc3 = ["dog", "likes", "meat"]

# Calculate WMD
distance1 = model.wv.wmdistance(doc1, doc2)
distance2 = model.wv.wmdistance(doc1, doc3)

print("WMD between Document 1 and Document 2:", distance1)
print("WMD between Document 1 and Document 3:", distance2)

if distance1 < distance2:
    print("Document 1 and Document 2 are more similar.")
else:
    print("Document 1 and Document 3 are more similar.")
```
