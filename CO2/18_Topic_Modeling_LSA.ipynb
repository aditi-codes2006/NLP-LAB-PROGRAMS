```python
# ============================================================
# PROGRAM 18
# TOPIC MODELING USING LSA
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# ------------------------------------------------------------
# Documents
# ------------------------------------------------------------

documents = [
    "football match team player goal stadium",
    "football player scored goal in the match",
    "team won the football championship",

    "machine learning artificial intelligence model",
    "deep learning neural network artificial intelligence",
    "machine learning algorithms train data",

    "python programming code software development",
    "software developer writes python code",
    "programming language python is popular"
]

# ------------------------------------------------------------
# TF-IDF
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(documents)

# ------------------------------------------------------------
# Apply SVD
# ------------------------------------------------------------

lsa = TruncatedSVD(
    n_components=3,
    random_state=42
)

lsa_matrix = lsa.fit_transform(X)

# ------------------------------------------------------------
# Vocabulary
# ------------------------------------------------------------

words = vectorizer.get_feature_names_out()

# ------------------------------------------------------------
# Display Topics
# ------------------------------------------------------------

for topic_number, component in enumerate(
    lsa.components_
):

    print("\nTopic", topic_number + 1)

    top_indexes = component.argsort()[-5:][::-1]

    top_words = [
        words[index]
        for index in top_indexes
    ]

    print("Top Words:", top_words)

# ------------------------------------------------------------
# Document-Topic Matrix
# ------------------------------------------------------------

print("\n================================")
print("DOCUMENT-TOPIC MATRIX")
print("================================")

print(lsa_matrix)
```
