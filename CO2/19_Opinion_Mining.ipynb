```python
# ============================================================
# PROGRAM 19
# OPINION MINING ON PRODUCT/SERVICE REVIEWS
# ============================================================

from textblob import TextBlob

# ------------------------------------------------------------
# Reviews Dataset
# ------------------------------------------------------------

reviews = [
    "The phone camera is excellent and the battery is amazing.",
    "The phone is very expensive and the battery is poor.",
    "The design is beautiful and performance is great.",
    "The service was terrible and the staff was rude.",
    "The product is okay, nothing special.",
    "Amazing quality and excellent customer service."
]

positive = 0
negative = 0
neutral = 0

# ------------------------------------------------------------
# Analyze reviews
# ------------------------------------------------------------

for review in reviews:

    blob = TextBlob(review)

    polarity = blob.sentiment.polarity

    if polarity > 0:
        opinion = "Positive"
        positive += 1

    elif polarity < 0:
        opinion = "Negative"
        negative += 1

    else:
        opinion = "Neutral"
        neutral += 1

    print("\nReview:", review)
    print("Polarity:", round(polarity, 3))
    print("Opinion:", opinion)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n================================")
print("OPINION MINING SUMMARY")
print("================================")

print("Positive Reviews:", positive)
print("Negative Reviews:", negative)
print("Neutral Reviews:", neutral)
```
