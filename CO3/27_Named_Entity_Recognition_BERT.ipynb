```python

from transformers import pipeline
import torch

# 1. Load the pre-trained BERT-based NER model
device = 0 if torch.cuda.is_available() else -1

ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple",
    device=device
)

# 2. Input text
text = """
Sundar Pichai is the CEO of Google.
Google is located in California, United States.
Microsoft was founded by Bill Gates and Paul Allen.
"""

# 3. Perform Named Entity Recognition
entities = ner(text)

# 4. Display the results
print("Input Text:")
print(text)

print("\nDetected Named Entities:\n")

for entity in entities:
    print("Entity:", entity["word"])
    print("Entity Type:", entity["entity_group"])
    print("Confidence:", round(entity["score"], 3))
    print("-" * 30)
```
