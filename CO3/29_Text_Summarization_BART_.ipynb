```python

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# 2. Load the pre-trained BART model
model_name = "sshleifer/distilbart-cnn-12-6"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

model.to(device)
model.eval()

# 3. Input text
text = """
Artificial intelligence is an important field of computer science.
It enables computers to perform tasks that normally require human
intelligence. Natural language processing helps computers understand
and generate human language.

NLP is used in machine translation, chatbots, sentiment analysis,
and text summarization. Transformer models such as BERT, GPT, and
BART have improved NLP applications. BERT is commonly used for
language understanding, GPT models are used for text generation,
and BART is useful for summarization.

Text summarization reduces long documents into shorter versions
while preserving important information. It helps students,
researchers, and professionals understand large amounts of
information quickly.
"""

# 4. Tokenize the input
inputs = tokenizer(
    text,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

# 5. Generate summary
with torch.no_grad():
    summary_ids = model.generate(
        **inputs,
        max_new_tokens=100,
        min_new_tokens=20,
        num_beams=4,
        do_sample=False
    )

# 6. Convert tokens into readable text
summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

# 7. Display results
print("\nORIGINAL TEXT:\n")
print(text)

print("\nGENERATED SUMMARY:\n")
print(summary)
```
