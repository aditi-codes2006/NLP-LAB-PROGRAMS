```python
# Tokenization of Sentences and Words using NLTK and spaCy

import nltk
import spacy

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize, word_tokenize

# Sample text
text = "Natural Language Processing is a branch of Artificial Intelligence. It helps computers understand human language."

print("Original Text:")
print(text)

# ---------------- NLTK ----------------

print("\n--- NLTK Sentence Tokenization ---")
sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)

print("\n--- NLTK Word Tokenization ---")
words = word_tokenize(text)
print(words)

# ---------------- spaCy ----------------

print("\n--- spaCy Tokenization ---")

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Process the text
doc = nlp(text)

print("\nSentences:")
for sentence in doc.sents:
    print(sentence.text)

print("\nWords:")
for token in doc:
    print(token.text)
```
